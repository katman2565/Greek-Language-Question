### HOCR FILES

```python
# Define the folder containing the files
data_folder = "SATHAS_JSON"

# Get a list of all .hocr files in the folder
hocr_files = [file for file in os.listdir(data_folder) if file.endswith(".hocr")]

# Initialize an empty list to store page data
data = []

# Process each .hocr file
for file in hocr_files:
    # Extract the page number from the file name
    page_number = int(file.split('_')[-1].split('.')[0])  # Assumes page number is at the end of the file name
    
    # Read the content of the HOCR file
    with open(os.path.join(data_folder, file), 'r', encoding='utf-8') as f:
        hocr_content = f.read()
    
    # Parse the content with BeautifulSoup
    soup = BeautifulSoup(hocr_content, 'html.parser')
    
    # Extract text content
    text = soup.get_text(separator=' ')  # Use separator=' ' to maintain spacing
    
    # Append the data to the list
    data.append({"Page_Number": page_number, "Text": text})

```

```python
# Create a pandas DataFrame
df = pd.DataFrame(data)

# Sort by page number for consistency
df = df.sort_values(by="Page_Number").reset_index(drop=True)

```

### Lets try to remove the footnotes before joining the text

```python
# 1. Pages with marked footnotes

marked_pages = [
    8, 10, 11, 12, 15, 17, 21, 24, 26, 27, 28, 29, 30, 31,
    33, 34, 35, 36, 37, 45, 46, 49, 50, 51, 52, 53, 55, 56,
    58, 59, 60, 61, 64, 74, 75, 76, 77, 78, 79, 81, 87, 89,
    90, 93, 95, 98, 99, 105, 109, 110, 112, 113, 114, 117,
    121, 123, 125, 127, 128, 129, 131, 147, 149, 176, 178,
    186, 187, 194, 203, 206, 216, 218, 221, 222, 223, 230,
    232, 235, 236, 243, 245, 247, 252, 257, 259, 265, 272,
    274, 283, 284, 288, 292, 294, 296, 297, 299, 300, 303,
    306, 307, 308, 309, 310
]

# 2. Regex for possible footnote markers
# detects forms like:# ( 1 ) # (1) # ( 12 ) # ( * ) # (*)

marker_re = re.compile(
    r"\(\s*(?:\*|\d{1,2})\s*\)"
)

# 3. Detect possible footnote areas

results = []

for page in marked_pages:

    # text at the specific page
    page_rows = df.loc[df["Page_Number"] == page, "Text"]

    if page_rows.empty:
        print(f"WARNING: Δεν βρέθηκε η σελίδα {page}")
        continue

    text = str(page_rows.iloc[0])

    # find all possible markers to the page
    matches = list(marker_re.finditer(text))

    # If no markers found keep the page for manual check 
    if not matches:
        results.append({
            "Page": page,
            "Marker": None,
            "Char_Position": None,
            "Relative_Position": None,
            "Possible_footnote": "NO MARKER DETECTED"
        })
        continue

    # each marker created possible footnote zone:
    # from the marker unitl the end of the page 
    for m in matches:

        start = m.start()

        relative_position = start / len(text)

        possible_footnote = text[start:].strip()

        results.append({
            "Page": page,
            "Marker": m.group(),
            "Char_Position": start,
            "Relative_Position": round(relative_position, 3),
            "Possible_footnote": possible_footnote
        })



```

```python
# 4. DataFrame with candidates

candidates = pd.DataFrame(results)

candidates = candidates.sort_values(
    ["Page", "Relative_Position"],
    na_position="last"
).reset_index(drop=True)


```

```python
artifact_words = r"""
(?:OCR|
ΠΑΝΕΠΙΣΤΗΜΙΟ|
ΑΝΕΠΙΣΤΗΜΙΟ|
ΕΠΑΝΕΠΙΣΤΗΜΙΟ|
ΝΕΠΙΣΤΗΜΙΟ|
ΙΝΝΕΠΙΣΤΗΜΙΟ|
ΙΑΝΕΠΙΣΤΗΜΙΟ|
ΚΡΗΤΗΣ|
ΝΕΟΕΛΛΗΝ\s*\.|
ΝΕΟΛΕΛΗΝ|
ΝΕΟΒΑΔΗΝ|
ΝΕΟΓΛΛΗΝ|
ΦΙΛΟΛΟΓΙΑΣ|
ΦΙΔΟΛΟΓΙΑΣ|
ΠΑΡΑΡΤΗΜΑ|
ΝΕΟΕΛΛΙΩΝ|
ΚΡΗΤΗ|
-ΝΕΠΙΣΤΗΜΙΟ|
ΠΑΡΑΡΙΟΜΑ|
ΜΕΛΙΣΤΗΜΙΟ|
ΠΑΡΑΡΤΙΑ|
ΝΕΟΕΛΛΙΙΝ|
ΦΜΟΛΟΓΙΑΣ|
ΠΟΥ|
ΝΕΟΕΛΛΗΝΙ(?![^\W\d_])|
ПАРАРТНА|
SHIHA|
ΝΕΟΕΔΔΗΝ|
ΙΝΝΕΠΙΣΤΗΜΙΟ)
"""

# 2. permitted between artefact words
# numbers, whitespace punctuation 
# e.g.OCR ΠΑΝΕΠΙΣΤΗΜΙΟ ΚΡΗΤΗΣ 316 ΝΕΟΕΛΛΗΝ . ΦΙΛΟΛΟΓΙΑΣ
# can be recognised as cluster.

junk_between = r"""[\s\d\.\,\:\;\*\-–—"'·•\\\/]*"""

# 3. Compile

artifact_cluster = re.compile(
    rf"""
    (?:
        {artifact_words}
        {junk_between}
    )+
    """,
    re.VERBOSE
)
```

```python
# 4. Remove artefacts only from the edges

def remove_edge_artifacts(text, edge_chars=180):

    text = str(text)

    start = text[:edge_chars]

    middle = (
        text[edge_chars:-edge_chars]
        if len(text) > 2 * edge_chars
        else ""
    )

    end = (
        text[-edge_chars:]
        if len(text) > edge_chars
        else ""
    )

    # remove artefact clusters at the begining 
    start = artifact_cluster.sub(" ", start)

    # remove artefact clusters at the end
    end = artifact_cluster.sub(" ", end)

    # unite
    cleaned = start + middle + end

    # Normalize whitespace
    cleaned = re.sub(r"\s+", " ", cleaned).strip()

    return cleaned
```

```python
def show_removed_edge_artifacts(text, edge_chars=180):
    text = str(text)

    found = []

    start = text[:edge_chars]
    end = text[-edge_chars:]

    for m in artifact_cluster.finditer(start):
        found.append(("START", m.group()))

    for m in artifact_cluster.finditer(end):
        found.append(("END", m.group()))

    return found
```

```python
def clean_text(text):
    """Conservative cleaning for OCR word-count preparation."""

    if not isinstance(text, str) or text.strip() == "":
        return text

    # 1. TRy to fix broken words due to hyphens
    # π.χ. πα- τριαρχείου -> πατριαρχείου
    text = re.sub(
        r'([^\W\d_]+)-\s+([^\W\d_]+)',
        r'\1\2',
        text,
        flags=re.UNICODE
    )

    # 2. Αφαιρούμε κενό πριν από βασική στίξη
    # π.χ. "πόλιν ." -> "πόλιν."
    text = re.sub(
        r'\s+([.,;:!?··])',
        r'\1',
        text
    )

    # 3. Πολλαπλά whitespace -> ένα κενό
    text = re.sub(
        r'\s+',
        ' ',
        text
    )

    # 4. Καθαρισμός αρχής/τέλους
    text = text.strip()

    return text
```

```python
df["Text_clean"] = (
    df["Text_no_foot_artifacts"]
    .apply(clean_text)
)
```

```python
# Define the function to find words split by a middle dot
def find_middle_dot_splits(text):
    """
    Find all instances of words split by a middle dot (·) in the text.
    """
    pattern = r'(\w+)\s·\s(\w+)'
    return re.findall(pattern, text)

# test df
data = {
    'Text': [
        "Ἐν τούτῳ ἐκδηλώνον · ταν οἱ προθέσεις τῶν Ἑλλήνων.",
        "Ἡ γλώσσα · θρησκεία ἦταν σημαντικές ἀξίες.",
        "Οὐδεμία λέξις περιέχει · τὸ σύμβολον.",
        "Οὐκ ἐφάνη · σὺν αὐτοῖς τὰ λόγια."
    ]
}
test = pd.DataFrame(data)

# Collect all middle dot splits from the 'Text' column
all_middle_dot_splits = []
for text in test['Text']:
    splits = find_middle_dot_splits(text)
    all_middle_dot_splits.extend(splits)  # Flatten the list

# Display the results in a clean format
print("Words split by '·':")
for part1, part2 in all_middle_dot_splits:
    print(f"{part1} · {part2}")
```

```python
import re
def clean_text2(text):
    """
    Cleans text while preserving essential punctuation:
    - Keeps: `.`, `?`, `;`, `!`, `:`
    - Removes unwanted symbols like `•`, `#`, `♦`, `’`, etc.
    """
    if not isinstance(text, str):
        return text  # Return unchanged if not a string

    # Remove unwanted special characters (DO NOT remove `?`, `;`, `:`, `!`, or `.`)
    text = re.sub(r"[\r\n\xa0$•)%*■®῾£^¥᾽»'♦’#“`«»(/\]]", " ", text)  # No `?`, `;`, `:`, `!`, `.`

    # Normalize spaces (removes extra spaces)
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
def extract_highlight_spans(page):

    annotations = []

    for annot in page.annots() or []:

        # ΜΟΝΟ πράσινα highlights / Highlight annotations
        # όχι Underline κτλ.
        if annot.type[1] != "Highlight":
            continue

        annotations.append(annot)

    # φυσική σειρά πάνω -> κάτω, αριστερά -> δεξιά
    annotations.sort(
        key=lambda a: (a.rect.y0, a.rect.x0)
    )

    spans = []

    for annot in annotations:

        text_parts = []
        verts = annot.vertices

        for i in range(0, len(verts), 4):

            quad = fitz.Quad(verts[i:i+4])
            rect = quad.rect

            txt = page.get_text(
                "text",
                clip=rect
            ).strip()

            if txt:
                text_parts.append(txt)

        full = " ".join(text_parts)

        # εδώ θέλουμε απλό whitespace normalization
        full = re.sub(r"\s+", " ", full).strip()

        if full:
            spans.append(full)

    return spans
```

```python
def boundary_snippets(text, n_words=4):

    words = text.split()

    start = " ".join(words[:n_words])
    end = " ".join(words[-n_words:])

    return start, end
```

```python
def normalize_for_match(text):

    text = text.lower()

    text = re.sub(r"\s+", " ", text)

    text = re.sub(
        r"[«»“”\"'’‘.,;:··!?()\[\]{}<>—–\-]",
        " ",
        text
    )

    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
def tokenize_with_offsets(text):

    return [
        {
            "text": m.group(),
            "start": m.start(),
            "end": m.end()
        }
        for m in re.finditer(
            r"[^\W_]+(?:['’᾿][^\W_]+)?",
            text,
            flags=re.UNICODE
        )
    ]
```

```python
def fuzzy_find_snippet_with_offsets(
    snippet,
    target_text,
    min_char=0,
    max_char=None
):

    snippet_norm = normalize_for_match(snippet)
    snippet_words = snippet_norm.split()

    tokens = tokenize_with_offsets(target_text)

    if max_char is None:
        max_char = len(target_text)

    allowed = [
        t for t in tokens
        if t["start"] >= min_char
        and t["end"] <= max_char
    ]

    best = None
    base_window = len(snippet_words)

    for window_words in range(
        max(3, base_window - 1),
        base_window + 2
    ):

        for i in range(
            len(allowed) - window_words + 1
        ):

            candidate_tokens = allowed[
                i:i + window_words
            ]

            raw_candidate = " ".join(
                t["text"]
                for t in candidate_tokens
            )

            candidate_norm = normalize_for_match(
                raw_candidate
            )

            score = fuzz.ratio(
                snippet_norm,
                candidate_norm
            )

            if best is None or score > best["score"]:

                best = {
                    "score": score,
                    "char_start":
                        candidate_tokens[0]["start"],
                    "char_end":
                        candidate_tokens[-1]["end"],
                    "candidate":
                        raw_candidate
                }

    return best
```

```python
def match_highlight_spans_to_ocr(
    spans,
    ocr_text,
    n_words=4
):

    results = []
    cursor = 0

    for i, span in enumerate(spans, 1):

        start_snippet, end_snippet = boundary_snippets(
            span,
            n_words=n_words
        )

        # Start highlight:
        # has to be after the previous span
        start_match = fuzzy_find_snippet_with_offsets(
            start_snippet,
            ocr_text,
            min_char=cursor
        )

        if start_match is None:
            continue

        # End highlight:
        # has to be after the start of the same span
        end_match = fuzzy_find_snippet_with_offsets(
            end_snippet,
            ocr_text,
            min_char=start_match["char_start"]
        )

        if end_match is None:
            continue

        results.append({
            "span": i,

            "start_snippet":
                start_snippet,

            "start_score":
                start_match["score"],

            "start_char":
                start_match["char_start"],

            "start_candidate":
                start_match["candidate"],

            "end_snippet":
                end_snippet,

            "end_score":
                end_match["score"],

            "end_char":
                end_match["char_end"],

            "end_candidate":
                end_match["candidate"]
        })

        cursor = end_match["char_end"]

    return results
```

```python
def snap_end_to_punctuation(text, pos):
    """
    Moves the end boundary slightly to the right
    so that it includes final punctuation and closing characters
    that belong to the highlighted segment.
    """

    allowed = set(" \t\n\r.,;:··!?)]}»")

    while pos < len(text) and text[pos] in allowed:
        pos += 1

    return pos
```

```python
def mark_nonhighlight_regions(
    ocr_text,
    results,
    marker="(?)"
):
    results = sorted(
        results,
        key=lambda x: x["start_char"]
    )

    boundaries = []

    # Before the first highlight
    first_start = results[0]["start_char"]

    gap = ocr_text[:first_start]

    if any(ch.isalpha() for ch in gap):
        boundaries.append(first_start)

    # between highlights
    for i in range(len(results) - 1):

        gap_start = results[i]["end_char"]
        gap_end = results[i + 1]["start_char"]

        gap = ocr_text[gap_start:gap_end]

        if any(ch.isalpha() for ch in gap):
            boundaries.extend([
                gap_start,
                gap_end
            ])

    # after the final highlight
    last_end = results[-1]["end_char"]

    gap = ocr_text[last_end:]

    if any(ch.isalpha() for ch in gap):
        boundaries.append(last_end)

    # put markers from right to left
    text = ocr_text

    for pos in sorted(boundaries, reverse=True):
        text = (
            text[:pos]
            + " (?) "
            + text[pos:]
        )

    return text
```

```python
def has_words(text):
    """True if the segment contains at least one letter."""
    return bool(re.search(r"[^\W\d_]", text, flags=re.UNICODE))


def build_page_segments(text, matches):
    """
    Returns segments:
    (start, end, state)
    
    state = 'sathas' for highlighted text
            'quote'  for non highlighted text 
    """

    matches = sorted(
        matches,
        key=lambda x: x["start_char"]
    )

    # if no highlight the whole page is a quote
    
    if not matches:
        if has_words(text):
            return [(0, len(text), "quote")]
        else:
            return []

    segments = []
    cursor = 0

    for m in matches:

        start = m["start_char"]
        end = m["end_char"]

        # empty spave begfore the highlight
        if start > cursor:
            gap = text[cursor:start]

            if has_words(gap):
                segments.append(
                    (cursor, start, "quote")
                )
            elif segments:
                # punctuation-only gap:
                # unite with the previous segment 
                prev_start, _, prev_state = segments[-1]
                segments[-1] = (
                    prev_start,
                    start,
                    prev_state
                )

        # highlighted = Sathas voice 
        segments.append(
            (start, end, "sathas")
        )

        cursor = end

    # text after the final highlight
    if cursor < len(text):

        gap = text[cursor:]

        if has_words(gap):
            segments.append(
                (cursor, len(text), "quote")
            )
        elif segments:
            prev_start, _, prev_state = segments[-1]
            segments[-1] = (
                prev_start,
                len(text),
                prev_state
            )

    return segments
```

```python
marked_pages = {}

previous_state = None

for page_number in df["Page_Number"]:

    text = df.loc[
        df["Page_Number"] == page_number,
        "Text_for_marking"
    ].iloc[0]

    page_matches = (
        matches_df[
            matches_df["Page_Number"] == page_number
        ]
        .to_dict("records")
    )

    segments = build_page_segments(
        text,
        page_matches
    )

    insertions = []

    for start, end, state in segments:

        # Marker only when it changes:
        # Sathas -> quote
        # or quote -> Sathas
        if previous_state is not None and state != previous_state:
            insertions.append(start)

        # if the book begins with quote
        elif previous_state is None and state == "quote":
            insertions.append(start)

        previous_state = state

    # insert from right to left 
    marked = text

    for pos in sorted(set(insertions), reverse=True):
        marked = (
            marked[:pos]
            + " (?) "
            + marked[pos:]
        )

    marked_pages[page_number] = marked
```

```python
import re
import unicodedata

def conservative_clean(text):

    if not isinstance(text, str):
        return ""

    # 1. Unicode normalization
    # NFC keeps polytonic and greek « »
    text = unicodedata.normalize("NFC", text)

    # 2. Non-breaking spaces -> normal spaces
    text = text.replace("\u00A0", " ")

    # 3. Tabs -> spaces
    text = text.replace("\t", " ")

    # 4. Multiple spaces -> one space
    # does not touch punctuation or quotation marks
    text = re.sub(r"[ ]{2,}", " ", text)

    # 5. remove spaces before basic puntuation marks 
    # e.g. "λέξη ." -> "λέξη."
    text = re.sub(r"\s+([.,;:!?])", r"\1", text)

    # 6. remove spaces within greek quotation marks
    # « κείμενο » -> «κείμενο»
    text = re.sub(r"«\s+", "«", text)
    text = re.sub(r"\s+»", "»", text)

    # 7. normalize empty lines
    text = re.sub(r"\n[ \t]*\n+", "\n", text)

    return text.strip()

```

```python
import unicodedata
import re

def convert_polytonic_to_simple(text):
    """
    Removes (δασεία and ψιλή) and converts περισπωμένη to oξεία.
    
    Args:
        text (str): Polytonic Greek text.
        
    Returns:
        str: Simplified text with πνευματα removed and περισπωμένη converted to ὀξεία.
    """
    # Normalize the text to NFD (decomposed form)
    text = unicodedata.normalize('NFD', text)
    
    # Remove breathing marks: δασεία (◌̔) and ψιλή (◌̓)
    text = re.sub(r'[\u0313\u0314]', '', text)  # Removes U+0313 (ψιλή) and U+0314 (δασεία)
    
    # Replace περισπωμένη (◌̃) with ὀξεία (◌́)
    text = re.sub(r'\u0342', '\u0301', text)  # Replace U+0342 (περισπωμένη) with U+0301 (ὀξεία)
    
    # Normalize back to NFC (composed form)
    text = unicodedata.normalize('NFC', text)
    
    return text
```

```python
def convert_polytonic_to_monotonic(text):
    """
    Convert Greek text from polytonic to monotonic orthography.
    Handle characters like \u1fc4 and replace with monotonic equivalents.
    """
    # Normalize to decomposed form (NFD)
    text = unicodedata.normalize('NFD', text)
    
    # Replace \u1fc4 (ᾴ) with άι
    text = text.replace('\u1fc4', 'ά')
    
    # Remove breathing marks: δασεία (◌̔) and ψιλή (◌̓)
    text = re.sub(r'[\u0313\u0314]', '', text)
    
    # Replace grave accent (◌̀) with acute accent (◌́)
    text = text.replace('\u0300', '\u0301')
    
    # Normalize back to composed form (NFC)
    text = unicodedata.normalize('NFC', text)
    
    return text
```

### Manual inspection and correction of 'broken' words

```python
def get_partial_context(text, target, window=5):
    """
    Finds the context around a target substring in the text, 
    including cases where words are split.

    Parameters:
    text (str): The text to search.
    target (str): The specific substring to find (partial match).
    window (int): The number of words before and after the target.

    Returns:
    list: A list of contexts where the target substring appears.
    """
    # Regex to match the target substring and surrounding words
    pattern = rf'((?:\S+\s+){{0,{window}}})(\S*{re.escape(target)}\S*)(?:\s+(\S+)){{0,{window}}}'
    
    # Find all matches
    matches = re.finditer(pattern, text)
    
    # Collect the contexts
    contexts = []
    for match in matches:
        before = match.group(1).strip()  # Words before the target
        target_word = match.group(2).strip()  # Word containing the target
        after = match.group(3) or ''  # Words after the target (if any)
        contexts.append(f"{before} {target_word} {after}".strip())
    
    return contexts

# Example 
text = "Έξω απ αυτή χτίζουμε καί θά χτί ζουμε στήν άμμο ."

# Specify the target substring
target = "χτί "

# Get the contexts for the target
contexts = get_partial_context(text, target, window=5)

# Print the results
print("Contexts found:")
for context in contexts:
    print(context)
```

```python
def pre_correct_stopwords_safe_case_sensitive(text, correction_dict):
    """
    Replaces stopwords with de-accented versions, preserving case.
    """
    # Compile a pattern for word boundaries and dictionary keys
    pattern = re.compile(r'\b(' + '|'.join(map(re.escape, correction_dict.keys())) + r')\b')
    
    def replace_with_case(match):
        """
        Replaces the matched word with the corresponding de-accented version,
        preserving the original case.
        """
        original_word = match.group(0)
        replacement_word = correction_dict[original_word.lower()]  # Get the de-accented version

        # Check if the original word was all uppercase or title case
        if original_word.isupper():
            return replacement_word.upper()  # Replace with uppercase
        elif original_word.istitle():
            return replacement_word.title()  # Replace with title case
        else:
            return replacement_word  # Replace with lowercase

    # Replace using the custom replacement function
    return pattern.sub(replace_with_case, text)
```
